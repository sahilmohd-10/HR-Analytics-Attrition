```python

df.drop_duplicates(inplace=True)

df['Attrition'] = df['Attrition'].map({'Yes':1, 'No':0})

print(df.isnull().sum())
```

```python
# Satisfaction Index (key feature)
df['Satisfaction_Index'] = (
    df['JobSatisfaction'] +
    df['EnvironmentSatisfaction'] +
    df['RelationshipSatisfaction']
) / 3

# Tenure Bands
df['Tenure_Band'] = pd.cut(
    df['YearsAtCompany'],
    bins=[0,2,5,10,40],
    labels=['0-2','3-5','6-10','10+']
)

# Overtime Flag
df['OverTime_Flag'] = df['OverTime'].apply(lambda x: 1 if x=='Yes' else 0)

# Salary Bands
df['Salary_Band'] = pd.cut(
    df['MonthlyIncome'],
    bins=[0,3000,6000,10000,20000],
    labels=['Low','Medium','High','Very High']
)
```

```python
attrition_rate = df['Attrition'].mean()
print("Attrition Rate:", attrition_rate)
```

```python
overtime_attrition = df.groupby('OverTime_Flag')['Attrition'].mean()
print(overtime_attrition)
```

```python
dept_attrition = df.groupby('Department')['Attrition'].mean().sort_values(ascending=False)
print(dept_attrition)
```

```python
salary_attrition = df.groupby('Salary_Band')['Attrition'].mean()
print(salary_attrition)
```

```python
satisfaction_attrition = df.groupby('Satisfaction_Index')['Attrition'].mean()
print(satisfaction_attrition)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Attrition count
sns.countplot(x='Attrition', data=df)
plt.title("Attrition Distribution")
plt.show()

# Overtime impact
sns.barplot(x='OverTime', y='Attrition', data=df)
plt.title("Overtime vs Attrition")
plt.show()

# Salary vs Attrition
sns.barplot(x='Salary_Band', y='Attrition', data=df)
plt.title("Salary vs Attrition")
plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

X = df[['Age','MonthlyIncome','OverTime_Flag','Satisfaction_Index']]
y = df['Attrition']

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)

model = LogisticRegression()
model.fit(X_train, y_train)

print("Accuracy:", model.score(X_test, y_test))
```
